```python
%sql
SELECT
  ROUND(SUM(price), 2)       AS total_vendas,
  COUNT(DISTINCT order_id)   AS pedidos
FROM workspace.olist_gold.fact_order_items
```

```python
%sql
SELECT
  p.category_en          AS categoria,
  ROUND(SUM(f.price), 2) AS vendas
FROM workspace.olist_gold.fact_order_items AS f
JOIN workspace.olist_gold.dim_products AS p ON f.product_id = p.product_id
GROUP BY p.category_en
ORDER BY vendas DESC
LIMIT 5
```

```python
%sql
SELECT
  c.state                AS estado,
  ROUND(SUM(f.price), 2) AS vendas
FROM workspace.olist_gold.fact_order_items AS f
JOIN workspace.olist_gold.dim_customers AS c ON f.customer_id = c.customer_id
GROUP BY c.state
ORDER BY vendas DESC
LIMIT 1
```

```python
%sql
SELECT
  ROUND(100 * SUM(CASE WHEN c.region = 'Sudeste' THEN f.price END) / SUM(f.price), 1) AS pct_sudeste
FROM workspace.olist_gold.fact_order_items AS f
JOIN workspace.olist_gold.dim_customers AS c ON f.customer_id = c.customer_id
```

```python
%sql
SELECT
  d.year_month           AS mes,
  ROUND(SUM(f.price), 2) AS vendas
FROM workspace.olist_gold.fact_order_items AS f
JOIN workspace.olist_gold.dim_date AS d ON f.purchase_date = d.full_date
GROUP BY d.year_month
ORDER BY vendas DESC
LIMIT 1
```

```python
%sql
SELECT
  COUNT_IF(is_late)                    AS itens_atrasados,
  COUNT_IF(delivered_date IS NOT NULL) AS itens_entregues,
  ROUND(100 * COUNT_IF(is_late) / COUNT_IF(delivered_date IS NOT NULL), 1) AS pct_atraso
FROM workspace.olist_gold.fact_order_items
```

```python
%sql
SELECT
  is_late                      AS atrasou,
  ROUND(AVG(review_score), 2)  AS nota_media,
  COUNT(review_score)          AS avaliacoes
FROM workspace.olist_gold.fact_order_items
WHERE delivered_date IS NOT NULL
GROUP BY is_late
ORDER BY is_late
```

```python
%sql
SELECT
  main_payment_type        AS tipo_pagamento,
  COUNT(DISTINCT order_id) AS pedidos
FROM workspace.olist_gold.fact_order_items
GROUP BY main_payment_type
ORDER BY pedidos DESC
LIMIT 1
```

```python
%sql
SELECT
  ROUND(SUM(price) / COUNT(DISTINCT order_id), 2) AS ticket_medio
FROM workspace.olist_gold.fact_order_items
```

```python
%sql
SELECT
  seller_id,
  ROUND(SUM(price), 2) AS vendas
FROM workspace.olist_gold.fact_order_items
GROUP BY seller_id
ORDER BY vendas DESC
LIMIT 3
```

```python
%sql
SELECT
  c.state                        AS estado,
  ROUND(AVG(f.delivery_days), 1) AS media_dias,
  COUNT(*)                       AS itens_entregues
FROM workspace.olist_gold.fact_order_items AS f
JOIN workspace.olist_gold.dim_customers AS c ON f.customer_id = c.customer_id
WHERE f.delivered_date IS NOT NULL
GROUP BY c.state
HAVING COUNT(*) >= 100
ORDER BY media_dias DESC
LIMIT 3
```
